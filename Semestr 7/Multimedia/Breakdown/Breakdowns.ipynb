```python
import numpy as np 
import matplotlib.pyplot as plt 

harmonics = 16
point_size = 100
amplitude = 2
zoom = 1

points = np.zeros(point_size)
```

```python
for i in range(harmonics):
    if i==0:
        continue
    Ak = (amplitude * 4) / (np.pi * i)
    for j in range(point_size):
        points[j] = points[j] + Ak * np.sin(2 * np.pi * i * j / point_size)
```

```python
for i in range(harmonics):
    if i==0:
        continue
    Ak = (2*point_size)/(np.pi**2) * (1-(-1)**i)/(i**2)
    for j in range(point_size):
        points[j] = points[j] + Ak * np.cos(2 * np.pi * i * j / point_size)
```

```python
t = range(point_size)
plt.subplot(2, 1, 1)
plt.plot(t, points)
plt.grid()

y = np.fft.fft(points, point_size)
yy = np.sqrt(y * np.conjugate(y)) / (point_size / 2)
yy = yy[:len(yy)//2]
plt.subplot(2,1,2)
plt.bar(t[:len(t)//2],yy)
plt.grid()

plt.show()
```
