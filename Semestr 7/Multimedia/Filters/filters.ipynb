```python
import numpy as np
import matplotlib.pyplot as plt

M = int(input("Size of output: "))
fc = int(input("Cut-out frequency: "))
f = int(input("Signal frequency"))
N = int(input("Signal point count: "))

z = np.zeros(M)
```

```python
for k in range(M):
    if(k - (M)/2) < 0:
        z[k]=(np.sin(2*np.pi*fc*(k-M/2)))/(k-M/2)
    elif (k-(M/2)) == 0:
        z[k] = 2*np.pi*fc
    elif (k-(M/2)) > 0:
        z[k]=(np.sin(2*np.pi*fc*(k-M/2)))/(k-M/2)

plt.plot(z)
```
