```python
import numpy as np
import matplotlib.pyplot as plt
import math

periods = 2
series_precision = 64
point_count = 1000
```

```python
# Sinus approx
x_range = np.array(range(point_count))
sin_range = np.zeros(point_count)
for i in range(point_count):
    if i == 0:
        continue
    angle = i * np.pi / point_count * 2 * periods
    y = 0
    for k in range(series_precision):
        y += (-1) ** k * (angle ** (2 * k + 1)) / (math.factorial(2 * k + 1))
    sin_range[i] = y
```

```python
# Cosinus approx
x_range = np.array(range(point_count))
cos_range = np.zeros(point_count)
for i in range(point_count):
    if i == 0:
        continue
    angle = i * np.pi / point_count * 2 * periods
    y = 0
    for k in range(series_precision):
        y += ((-1) ** k * (angle ** (2 * k))) / (math.factorial(2 * k))
    cos_range[i] = y
cos_range[0] = 1
```

```python
plt.plot(sin_range)
plt.plot(cos_range)
plt.show()
```
